```python
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
articles_path='../data/articles.csv'
customer_path='../data/customers.csv'
transaction_path='../data/transactions_train.csv'
```

```python
articles=pd.read_csv(articles_path)
customers=pd.read_csv(customer_path)
transactions=pd.read_csv(transaction_path)
```

# articles

```python
articles.info()
```

这张表包含了 H&M 所有商品及其详细信息，如产品类型、颜色、产品组以及其他特征。

**商品数据说明：**

* **article_id**：每件商品的唯一标识符。
* **product_code, prod_name**：款式的唯一标识符及其名称。
* **product_type, product_type_name**：产品代码所属的类别及其名称。
* **graphical_appearance_no, graphical_appearance_name**：图案分组及其名称。
* **colour_group_code, colour_group_name**：颜色分组及其名称。
* **perceived_colour_value_id, perceived_colour_value_name, perceived_colour_master_id, perceived_colour_master_name**：附加的颜色信息。
* **department_no, department_name**：每个部门的唯一标识符及其名称。
* **index_code, index_name**：每个索引的唯一标识符及其名称。
* **index_group_no, index_group_name**：索引分组及其名称。
* **section_no, section_name**：每个分区的唯一标识符及其名称。
* **garment_group_no, garment_group_name**：服装分组的唯一标识符及其名称。
* **detail_desc**：商品详细描述。


```python
articles.head()
```

```python
# 删除无用的数据
articles=articles[[column for column in articles.columns if column.split('_')[-1]!='name']]
articles.drop(columns=['detail_desc'],inplace=True)
```

```python
articles.head()
```

```python
# 观察独特的数值
for col in articles.columns:
    if 'no'in col or 'code' in col:
        un_n=articles[col].nunique()
        print(f'n of unique {col}: {un_n}')
```

```python
articles.to_parquet('../data/articles.parquet')
```

# customer

```python
customers.info()
```

* **customer_id**：每位客户的唯一标识符
* **FN**：1 或缺失（可能表示客户是否填写了某些信息）
* **Active**：1 或缺失（可能表示客户是否活跃）
* **club_member_status**：会员状态
* **fashion_news_frequency**：H&M 向客户发送时尚资讯的频率
* **age**：当前年龄
* **postal_code**：客户的邮政编码


```python
customers.head()
```

```python
customers['Active'].value_counts()
```

```python
customers['FN'].value_counts()
```

```python
customers['fashion_news_frequency'].value_counts()
```

```python
customers['fashion_news_frequency'].value_counts()
```

```python
# 填充缺失值
customers['Active'].fillna(0,inplace=True)
customers['FN'].fillna(0,inplace=True)
customers['FN']=customers['FN'].astype('uint8')
customers['Active']=customers['Active'].astype('uint8')

customers['age'].fillna(customers['age'].mean(),inplace=True)
customers['club_member_status'].fillna(customers['club_member_status'].mode()[0],inplace=True)
customers['fashion_news_frequency'].fillna(customers['fashion_news_frequency'].mode()[0],inplace=True)
```

```python
customers.info()
```

```python
customers.head()
```

```python
customers['age'].hist()
```

```python
bins=[0,40,60,float('inf')]
labels=[0,1,2]
customers['age']=pd.cut(customers['age'],bins=bins,labels=labels)
customers.head()
```

```python
customers=pd.get_dummies(customers,columns=['fashion_news_frequency','club_member_status','age'])
```

```python
customers.head()
```

```python
customers.drop(['postal_code'],inplace=True,axis=1)
```

```python
customers.to_parquet('../data/customers.parquet')
```

# Transactions

```python
transactions.info()
```

交易数据说明：

* **t_dat**：每笔交易的日期（原文可能有误，通常是日期，而不是唯一标识符）
* **customer_id**：每位客户的唯一标识符（对应 customers 表）
* **article_id**：每件商品的唯一标识符（对应 articles 表）
* **price**：购买价格
* **sales_channel_id**：销售渠道，1 或 2


```python
transactions['sales_channel_id']=transactions['sales_channel_id'].astype('uint8')
```

```python
transactions['t_dat']=pd.to_datetime(transactions['t_dat'])
transactions['time']=transactions['t_dat'].apply(lambda x: x.timestamp())
```

```python
transactions.drop(['t_dat'],inplace=True,axis=1)
```
